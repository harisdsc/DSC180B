```python
import pandas as pd
import numpy as np
consDF = pd.read_parquet("/uss/hdsi-prismdata/q2-ucsd-consDF.pqt")
consDF = consDF.drop(columns = ["credit_score"])
consDF = consDF.dropna()
acctDF = pd.read_parquet("/uss/hdsi-prismdata/q2-ucsd-acctDF.pqt")
trxnDF = pd.read_parquet("/uss/hdsi-prismdata/q2-ucsd-trxnDF.pqt")
catmap = pd.read_csv("/uss/hdsi-prismdata/q2-ucsd-cat-map.csv")
# 1. Filter the DataFrame into positive (1.0) and negative (0.0) cases
positive_cases = consDF[consDF['DQ_TARGET'] == 1.0]
negative_cases = consDF[consDF['DQ_TARGET'] == 0.0]

# 2. Sample 20 IDs from each (ensuring we don't exceed available rows)
# Use 'prism_consumer_id' column to create the list
pos_id_list = positive_cases['prism_consumer_id'].sample(n=20, random_state=42).tolist()
neg_id_list = negative_cases['prism_consumer_id'].sample(n=20, random_state=42).tolist()

# 3. View the results
print("Positive IDs:", pos_id_list)
print("Negative IDs:", neg_id_list)
```

```python
# 1. Identify which users have at least one BNPL transaction
bnpl_users = trxnDF[trxnDF['category'] == 35]['prism_consumer_id'].unique()

# 2. Check how many of your sampled Positive IDs are in that BNPL list
pos_bnpl_count = pd.Series(pos_id_list).isin(bnpl_users).sum()

# 3. Check how many of your sampled Negative IDs are in that BNPL list
neg_bnpl_count = pd.Series(neg_id_list).isin(bnpl_users).sum()

# Display results
print(f"Positive Sample: {pos_bnpl_count} out of {len(pos_id_list)} used BNPL ({pos_bnpl_count/len(pos_id_list):.1%})")
print(f"Negative Sample: {neg_bnpl_count} out of {len(neg_id_list)} used BNPL ({neg_bnpl_count/len(neg_id_list):.1%})")
```

```python
def check_paycheck_regularity(user_id, df):
    # 1. Filter for Paycheck categories (3 and 5) and CREDIT only
    pay_df = df[
        (df['prism_consumer_id'] == user_id) & 
        (df['category'] == 3)
    ].sort_values('posted_date')
    
    if len(pay_df) < 3:
        return "Insufficient Data", 0
    
    # 2. Calculate days between consecutive paychecks
    pay_df['days_diff'] = pay_df['posted_date'].diff().dt.days
    deltas = pay_df['days_diff'].dropna()
    
    # 3. Heuristic: Check common frequencies
    avg_gap = deltas.mean()
    std_gap = deltas.std() # Low standard deviation = high regularity
    
    # 4. Determine Pattern
    pattern = "Irregular"
    if 6 <= avg_gap <= 8: pattern = "Weekly"
    elif 13 <= avg_gap <= 16: pattern = "Bi-Weekly"
    elif 27 <= avg_gap <= 32: pattern = "Monthly"
    
    # Regularity Score (Coefficient of Variation)
    # Lower is better; it means the paychecks arrive exactly the same number of days apart
    regularity_score = std_gap / avg_gap if avg_gap != 0 else 1
    
    return pattern, regularity_score

# Apply to your samples
regularity_results = []
for uid in (pos_id_list + neg_id_list):
    pattern, score = check_paycheck_regularity(uid, trxnDF)
    regularity_results.append({
        'prism_consumer_id': uid,
        'group': 'Positive' if uid in pos_id_list else 'Negative',
        'pay_pattern': pattern,
        'stability_score': round(score, 3)
    })

reg_df = pd.DataFrame(regularity_results)

# Summary View
print("--- Paycheck Regularity by Group ---")
print(reg_df.groupby(['group', 'pay_pattern']).size().unstack(fill_value=0))
print("\n--- Average Stability Score (Lower is more stable) ---")
print(reg_df.groupby('group')['stability_score'].mean())
```

```python
import numpy as np
import pandas as pd

def calculate_stability(df, user_ids, cat_id):
    """Calculates the average stability score for a list of users for a specific category."""
    scores = []
    for uid in user_ids:
        # Filter for specific user and category
        temp_df = df[(df['prism_consumer_id'] == uid) & (df['category'] == cat_id)].sort_values('posted_date')
        
        if len(temp_df) >= 3:
            deltas = temp_df['posted_date'].diff().dt.days.dropna()
            avg_gap = deltas.mean()
            std_gap = deltas.std()
            # Coefficient of Variation: Lower = More Regular
            score = std_gap / avg_gap if avg_gap > 0 else 1.0
            scores.append(score)
        else:
            # Not enough data points to establish a pattern
            scores.append(np.nan) 
            
    return np.nanmean(scores) if scores else np.nan

# 1. Get list of all unique categories
all_categories = trxnDF['category'].unique()
discrepancy_report = []

# 2. Loop through every category to find discrepancies
for cat_id in all_categories:
    pos_stability = calculate_stability(trxnDF, pos_id_list, cat_id)
    neg_stability = calculate_stability(trxnDF, neg_id_list, cat_id)
    
    if not np.isnan(pos_stability) and not np.isnan(neg_stability):
        discrepancy_report.append({
            'category_id': cat_id,
            'pos_avg_stability': round(pos_stability, 4),
            'neg_avg_stability': round(neg_stability, 4),
            'discrepancy': round(abs(pos_stability - neg_stability), 4)
        })

# 3. Rank by Discrepancy
rank_df = pd.DataFrame(discrepancy_report)

# Map names from your catmap
cat_names = dict(zip(catmap['category_id'], catmap['category']))
rank_df['category_name'] = rank_df['category_id'].map(cat_names)

# Sort: Highest discrepancy at the top
ranked_results = rank_df[['category_name', 'pos_avg_stability', 'neg_avg_stability', 'discrepancy']]
print(ranked_results.sort_values(by='discrepancy', ascending=False).head(10))
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import roc_auc_score

# --- 1. Aggregated Stability Feature Engineering ---
# We calculate Mean Gap and Std Gap per user per category
target_categories = [3, 12, 31, 32, 34, 35, 36] # Paycheck, Loan, Bills, Mortgage, Rent, BNPL, Auto Loan

def get_temporal_features(group):
    if len(group) < 3:
        return pd.Series({'avg_gap': 0, 'std_gap': 0})
    deltas = group['posted_date'].sort_values().diff().dt.days.dropna()
    return pd.Series({'avg_gap': deltas.mean(), 'std_gap': deltas.std()})

# Calculate features across trxnDF
stability_stats = trxnDF[trxnDF['category'].isin(target_categories)].groupby(
    ['prism_consumer_id', 'category']
).apply(get_temporal_features).unstack()

# Flatten multi-index columns: 'avg_gap_3', 'std_gap_3', etc.
stability_stats.columns = [f"{col}_{cat}" for col, cat in stability_stats.columns]

# --- 2. Feature Merging & BNPL Logic ---
# Create BNPL frequency feature
bnpl_counts = trxnDF[trxnDF['category'] == 35].groupby('prism_consumer_id').size()
consDF['bnpl_count'] = consDF['prism_consumer_id'].map(bnpl_counts).fillna(0)
consDF['bnpl_gt_5'] = (consDF['bnpl_count'] > 5).astype(int)

# Combine into final feature matrix
X_df = consDF[['prism_consumer_id', 'bnpl_gt_5', 'DQ_TARGET']].merge(
    stability_stats, on='prism_consumer_id', how='left'
).fillna(0)

# Select all features except ID and Target
features = [col for col in X_df.columns if col not in ['prism_consumer_id', 'DQ_TARGET']]
X = X_df[features].values
y = X_df['DQ_TARGET'].values

# Standardize: Critical since gaps and std_devs vary in scale
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Convert to Tensors
X_train_t = torch.FloatTensor(X_train)
y_train_t = torch.FloatTensor(y_train).view(-1, 1)
X_test_t = torch.FloatTensor(X_test)

# --- 3. 2-Hidden-Layer Neural Network ---
class StabilityNet(nn.Module):
    def __init__(self, input_dim):
        super(StabilityNet, self).__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, 32), # Wider first layer for raw feature complexity
            nn.ReLU(),
            nn.Linear(32, 16),        # Second hidden layer
            nn.ReLU(),
            nn.Linear(16, 1),
            nn.Sigmoid()
        )
    def forward(self, x):
        return self.net(x)

# --- 4. Training & Evaluation ---
device = torch.device("cuda" if torch.cuda.is_available() else "mps" if torch.backends.mps.is_available() else "cpu")
print(f"Training on: {device}")

# 2. Move your model to that device
model = StabilityNet(input_dim=len(features)).to(device)

# 3. Move your tensors to that device during the training loop
X_train_t = torch.FloatTensor(X_train).to(device)
y_train_t = torch.FloatTensor(y_train).view(-1, 1).to(device)
optimizer = optim.Adam(model.parameters(), lr=0.005)
criterion = nn.BCELoss()

for epoch in range(200):
    model.train()
    optimizer.zero_grad()
    loss = criterion(model(X_train_t), y_train_t)
    loss.backward()
    optimizer.step()

# ROC AUC Performance
model.eval()
with torch.no_grad():
    y_probs = model(X_test_t).numpy()
    auc = roc_auc_score(y_test, y_probs)

print(f"Features: Average Gap, Std Dev, and BNPL Threshold")
print(f"Total Features Trained: {len(features)}")
print(f"Final ROC AUC Score: {auc:.4f}")
```
