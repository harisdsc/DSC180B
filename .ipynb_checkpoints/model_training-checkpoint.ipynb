```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, accuracy_score
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
from sklearn.model_selection import StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.feature_selection import SequentialFeatureSelector
```

```python
consDF = pd.read_parquet("/uss/hdsi-prismdata/q2-ucsd-consDF.pqt")
consDF = consDF.drop(columns = ["credit_score"])
consDF = consDF.dropna()
acctDF = pd.read_parquet("/uss/hdsi-prismdata/q2-ucsd-acctDF.pqt")
trxnDF = pd.read_parquet("/uss/hdsi-prismdata/q2-ucsd-trxnDF.pqt")
catmap = pd.read_csv("/uss/hdsi-prismdata/q2-ucsd-cat-map.csv")
acctDF = acctDF[acctDF["prism_consumer_id"].isin(consDF['prism_consumer_id'])]
acctDF["balance_date"] = pd.to_datetime(acctDF["balance_date"])
trxnDF = trxnDF[trxnDF["prism_consumer_id"].isin(consDF['prism_consumer_id'])]
trxnDF = trxnDF.drop_duplicates()
trxnDF["posted_date"] = pd.to_datetime(trxnDF["posted_date"])
trxnDF['posted_date'] = pd.to_datetime(trxnDF['posted_date'])
consDF['evaluation_date'] = pd.to_datetime(consDF['evaluation_date'])

# Merge evaluation_date into transaction DF
trxnDF = trxnDF.merge(
    consDF[['prism_consumer_id', 'evaluation_date']],
    on='prism_consumer_id',
    how='left'
)
acctDF = acctDF.merge(
    consDF[['prism_consumer_id', 'evaluation_date']],
    on='prism_consumer_id',
    how='left'
)
# Keep only transactions before or on evaluation_date
trxnDF = trxnDF[trxnDF['posted_date'] <= trxnDF['evaluation_date']].copy()
acctDF = acctDF[acctDF['balance_date'] <= acctDF['evaluation_date']].copy()
all_feats = pd.read_parquet("all_feats.parquet")
all_feats_shrink = pd.read_parquet("all_feats_shrink.parquet")
```

```python
zero_cols = (all_feats == 0).all()
all_feats = all_feats.loc[:, ~zero_cols]
min_nonzero = 1
num_cols = all_feats.select_dtypes(include="number").columns
nonzero_counts = (all_feats[num_cols] != 0).sum()

keep_cols = nonzero_counts[nonzero_counts >= min_nonzero].index
all_feats = all_feats[["prism_consumer_id"]+ keep_cols.tolist()]
```

## Logistic

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy="mean")
feature_cols = all_feats.columns.difference(["prism_consumer_id"])
all_feats[feature_cols] = imputer.fit_transform(
    all_feats[feature_cols]
)
all_feats = all_feats.merge(consDF[["prism_consumer_id", "DQ_TARGET"]],
    on="prism_consumer_id",
    how="inner" 
)
```

```python
all_feats = all_feats.drop(columns=["DQ_TARGET_x"])
```

```python
ID_COL = "prism_consumer_id"
TARGET = "DQ_TARGET_y"

X = all_feats.drop(columns=[ID_COL, TARGET])
y = all_feats[TARGET].astype(int)

X = X.select_dtypes(include="number")
X = X.fillna(X.median())
```

```python
scaler = StandardScaler()
X_scaled = pd.DataFrame(
    scaler.fit_transform(X),
    columns=X.columns,
    index=X.index,
)

cv = StratifiedKFold(n_splits=8, shuffle=True, random_state=42)

coef_sum = np.zeros(X_scaled.shape[1])

for tr, va in cv.split(X_scaled, y):
    model = LogisticRegression(
        penalty="l1",
        solver="liblinear",
        C=0.1,
        max_iter=2000,
    )
    model.fit(X_scaled.iloc[tr], y.iloc[tr])
    coef_sum += np.abs(model.coef_[0])

importance = pd.Series(
    coef_sum / cv.get_n_splits(),
    index=X_scaled.columns
).sort_values(ascending=False)

top_50_features = importance.head(50).index.tolist()
top_50_features
```

```python
def add_engineered_features(X, trx_df, top_features, id_col='prism_consumer_id', date_col='posted_date'):
    """
    X: DataFrame with top 50 features (aggregated per consumer)
    trx_df: raw transaction DataFrame with columns [id_col, date_col, category, amount]
    top_features: list of top features
    Returns: X with additional engineered features
    """
    X_new = X.copy()

    # -------------------------
    # Trend / slope features
    # -------------------------
    for feat in top_features:
        if 'balance' in feat or 'ratio' in feat or 'spend' in feat:
            # Example: last period - previous period
            X_new[f'{feat}_slope'] = X_new[feat] - X_new.get(f'{feat}_prev', X_new[feat])
            
            # Rolling statistics (mocked if aggregated; replace with real time series if available)
            X_new[f'{feat}_rolling_mean'] = X_new[feat]  # placeholder: if you have time-series, compute rolling mean
            X_new[f'{feat}_rolling_std'] = 0  # placeholder

    # -------------------------
    # Frequency / count features
    # -------------------------
    # Count number of transactions per category in last 6 months
    cat_cols = [str(c) for c in trx_df['category'].unique() if str(c).startswith('cat')]
    txn_counts = trx_df.groupby([id_col, 'category'])[date_col].count().unstack(fill_value=0)
    txn_counts.columns = [f'{col}_txn_count_6m' for col in txn_counts.columns]

    X_new = X_new.merge(txn_counts, left_index=True, right_index=True, how='left')
    X_new.fillna(0, inplace=True)

    # -------------------------
    # Interaction features
    # -------------------------
    key_ratios = [f for f in top_features if 'ratio' in f]
    key_balances = [f for f in top_features if 'balance' in f]
    
    for r in key_ratios:
        for b in key_balances:
            X_new[f'{r}_x_{b}'] = X_new[r] * X_new[b]

    # -------------------------
    # Volatility features
    # -------------------------
    # If you have daily balance, you could compute std over last 30d/90d
    # Example placeholders:
    for b in key_balances:
        X_new[f'{b}_volatility'] = 0  # placeholder

    return X_new

# Usage:
X_fe = add_engineered_features(all_feats, trxnDF, top_50_features)

# Now X_fe has trend, interaction, and frequency features ready for training
print("Engineered feature set shape:", X_fe.shape)
```

```python
ID_COL = "prism_consumer_id"
TARGET = "DQ_TARGET_y"

X = X_fe.drop(columns=[ID_COL, TARGET])
y = X_fe[TARGET].astype(int)

X = X.select_dtypes(include="number")
X = X.fillna(X.median())
scaler = StandardScaler()
X_scaled = pd.DataFrame(
    scaler.fit_transform(X),
    columns=X.columns,
    index=X.index,
)

cv = StratifiedKFold(n_splits=8, shuffle=True, random_state=42)

coef_sum = np.zeros(X_scaled.shape[1])

for tr, va in cv.split(X_scaled, y):
    model = LogisticRegression(
        penalty="l1",
        solver="liblinear",
        C=0.1,
        max_iter=2000,
    )
    model.fit(X_scaled.iloc[tr], y.iloc[tr])
    coef_sum += np.abs(model.coef_[0])

importance = pd.Series(
    coef_sum / cv.get_n_splits(),
    index=X_scaled.columns
).sort_values(ascending=False)

top_50_features = importance.head(50).index.tolist()
top_50_features
```

```python
from sklearn.metrics import roc_auc_score
scaler = StandardScaler()
X_scaled = pd.DataFrame(
    scaler.fit_transform(X),
    columns=X.columns,
    index=X.index,
)

auc_scores = []
coef_store = []
cv = StratifiedKFold(n_splits=8, shuffle=True, random_state=42)
for tr, va in cv.split(X_scaled, y):
    model = LogisticRegression(
        penalty="l1",
        solver="liblinear",
        C=0.1,
        max_iter=2000,
    )
    model.fit(X_scaled.iloc[tr], y.iloc[tr])

    y_prob = model.predict_proba(X_scaled.iloc[va])[:, 1]
    auc = roc_auc_score(y.iloc[va], y_prob)
    auc_scores.append(auc)

    coef_store.append(np.abs(model.coef_[0]))

coef_store = np.array(coef_store)
auc_scores = np.array(auc_scores)

weighted_importance = (coef_store.T @ auc_scores) / auc_scores.sum()

importance_auc = pd.Series(
    weighted_importance,
    index=X_scaled.columns
).sort_values(ascending=False)

top_features_auc = importance_auc.head(50)
```

```python
top_features_auc = top_features_auc.index.tolist()
```

```python
X_top50 = X_fe[[ID_COL, TARGET] + top_50_features]
```

```python
import time
from sklearn.model_selection import GroupShuffleSplit
splitter = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, test_idx = next(splitter.split(X_top50, groups=X_top50['prism_consumer_id']))

X = X_top50[top_50_features]
y = X_top50['DQ_TARGET_y'].astype(int)

X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

# --- 6. Pipeline ---
# Note: n_jobs is technically deprecated in the init for some versions, 
# but mostly harmless. Removed to be safe.
clf = Pipeline(steps=[
    ('scaler', StandardScaler()), 
    ('classifier', LogisticRegression(random_state=42, 
                                      max_iter=1000, 
                                      class_weight='balanced', 
                                      solver='saga')) 
])

print("Training model...")
start_train = time.perf_counter()
clf.fit(X_train, y_train)
train_time = time.perf_counter() - start_train

# --- 7. Evaluate ---
start_score = time.perf_counter()
y_prob = clf.predict_proba(X_test)[:, 1]
score_time = time.perf_counter() - start_score
y_prob_train = clf.predict_proba(X_train)[:, 1]
roc_auc = roc_auc_score(y_test, y_prob)
roc_auc_train = roc_auc_score(y_train, y_prob_train)
latency_per_sample = score_time / len(X_test)
latency_per_sample_ms = latency_per_sample * 1_000
roc_auc, roc_auc_train
```

## XGBoost

```python
import xgboost as xgb
from tqdm.auto import tqdm
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold
```

```python
ID_COL = "prism_consumer_id"
TARGET = "DQ_TARGET_y"

X = all_feats.drop(columns=[ID_COL, TARGET])
y = all_feats[TARGET].astype(int)

X = X.select_dtypes(include="number")
X = X.fillna(X.median())
```

```python
import xgboost as xgb
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedKFold
cv = StratifiedKFold(
    n_splits=8,
    shuffle=True,
    random_state=42
)

feature_names = X.columns
n_features = len(feature_names)

importance_sum = np.zeros(n_features)
auc_scores = []
total_auc = 0.0

# -------------------------
# Cross-validated training
# -------------------------
for fold, (tr, va) in enumerate(cv.split(X, y), 1):

    model = xgb.XGBClassifier(
        objective="binary:logistic",
        eval_metric="auc",
        n_estimators=300,
        max_depth=4,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        min_child_weight=20,
        reg_alpha=1.0,
        reg_lambda=10.0,
        random_state=42,
        n_jobs=-1,
        tree_method="hist"
    )

    # Train
    model.fit(X.iloc[tr], y.iloc[tr])

    # Predict for AUC
    y_prob = model.predict_proba(X.iloc[va])[:, 1]
    auc = roc_auc_score(y.iloc[va], y_prob)

    auc_scores.append(auc)
    total_auc += auc

    # Gain-based importance
    fold_importance = model.get_booster().get_score(
        importance_type="gain"
    )

    # Normalize gain within fold
    total_gain = sum(fold_importance.values())

    if total_gain > 0:
        for i, f in enumerate(feature_names):
            importance_sum[i] += (
                fold_importance.get(f, 0.0) / total_gain
            ) * auc

    print(f"Fold {fold} AUC: {auc:.4f}")

# -------------------------
# Final feature importance
# -------------------------
importance = pd.Series(
    importance_sum / total_auc,
    index=feature_names
).sort_values(ascending=False)

# -------------------------
# Select top features
# -------------------------
TOP_K = 50
top_50_features = importance.head(TOP_K).index.tolist()

# -------------------------
# Diagnostics
# -------------------------
print("\nCV AUC Summary")
print(f"Mean AUC: {np.mean(auc_scores):.4f}")
print(f"Std AUC:  {np.std(auc_scores):.4f}")

print("\nTop 15 Features:")
print(importance.head(15))

top_50_features
```

```python
X_top50 = all_feats[[ID_COL, TARGET] + top_50_features]
```

```python
from sklearn.model_selection import GroupShuffleSplit

# 20% holdout TEST
gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
trainval_idx, test_idx = next(
    gss.split(all_feats, groups=all_feats['prism_consumer_id'])
)

X = X_top50[top_50_features]
y = X_top50['DQ_TARGET_y']

X_trainval, X_test = X.iloc[trainval_idx], X.iloc[test_idx]
y_trainval, y_test = y.iloc[trainval_idx], y.iloc[test_idx]
groups_trainval = all_feats_shrink.iloc[trainval_idx]['prism_consumer_id']

# 20% of trainval → validation (i.e. 64/16/20 split)
gss_inner = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, val_idx = next(
    gss_inner.split(X_trainval, groups=groups_trainval)
)

X_train, X_val = X_trainval.iloc[train_idx], X_trainval.iloc[val_idx]
y_train, y_val = y_trainval.iloc[train_idx], y_trainval.iloc[val_idx]

corr_matrix = X_train[top_50_features].corr().abs()
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
to_drop = [col for col in upper.columns if any(upper[col] > 0.9)]
X_train.drop(columns=to_drop, inplace=True)
X_val.drop(columns=to_drop, inplace=True)
X_test.drop(columns=to_drop, inplace=True)
```

```python
import xgboost as xgb
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedKFold
dtrain = xgb.DMatrix(X_train, label=y_train)
dval = xgb.DMatrix(X_val, label=y_val)
dtest = xgb.DMatrix(X_test, label=y_test)

params = {
    "objective": "binary:logistic",
    "eval_metric": "auc",
    "max_depth": 4,
    "learning_rate": 0.02,
    "subsample": 0.8,
    "colsample_bytree": 0.8,
    "min_child_weight": 20,
    "reg_alpha": 5,
    "reg_lambda": 20,
    "tree_method": "hist",
    "seed": 42
}

evallist = [(dtrain, 'train'), (dval, 'val')]

bst = xgb.train(
    params,
    dtrain,
    num_boost_round=1000,
    evals=evallist,
    early_stopping_rounds=50,
    verbose_eval=10
)

# Predict
y_val_pred = bst.predict(dval)
y_test_pred = bst.predict(dtest)

val_auc = roc_auc_score(y_val, y_val_pred)
test_auc = roc_auc_score(y_test, y_test_pred)

print(f"Validation AUC: {val_auc:.4f}")
print(f"Test AUC:       {test_auc:.4f}")

# Feature importance
importance_df = pd.DataFrame(
    bst.get_score(importance_type='gain').items(),
    columns=['feature', 'gain']
).sort_values('gain', ascending=False)

print("\nTop 15 Features (Final Model):")
print(importance_df.head(15))
```

```python
import time
from tqdm import tqdm
import joblib
from contextlib import contextmanager
@contextmanager
def tqdm_joblib(tqdm_object):
    class TqdmBatchCompletionCallback(joblib.parallel.BatchCompletionCallBack):
        def __call__(self, *args, **kwargs):
            tqdm_object.update(n=self.batch_size)
            return super().__call__(*args, **kwargs)

    old_callback = joblib.parallel.BatchCompletionCallBack
    joblib.parallel.BatchCompletionCallBack = TqdmBatchCompletionCallback
    try:
        yield tqdm_object
    finally:
        joblib.parallel.BatchCompletionCallBack = old_callback
        tqdm_object.close()
```

```python
from sklearn.model_selection import StratifiedKFold, RandomizedSearchCV
import xgboost as xgb
import time
from tqdm import tqdm
import joblib
from contextlib import contextmanager

scale_pos_weight = (y_train == 0).sum() / (y_train == 1).sum()

xgb_model = xgb.XGBClassifier(
    objective="binary:logistic",
    eval_metric="auc",
    scale_pos_weight=scale_pos_weight,
    random_state=42,
    n_jobs=-1
)

param_dist = {
    "n_estimators": [300, 500, 700],
    "max_depth": [2, 3, 4, 5],
    "learning_rate": [0.01, 0.02, 0.03],
    "subsample": [0.5, 0.6, 0.7],
    "colsample_bytree": [0.5, 0.6, 0.7],
    "min_child_weight": [10, 20, 50],
    "gamma": [0.5, 1, 2, 5],
    "reg_alpha": [0.1, 1, 5, 10],
    "reg_lambda": [5, 10, 20]
}

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

search = RandomizedSearchCV(
    estimator=xgb_model,
    param_distributions=param_dist,
    n_iter=50,
    scoring="roc_auc",
    cv=cv,
    n_jobs=-1,
    random_state=42
)

total_fits = 50 * cv.get_n_splits()

with tqdm_joblib(tqdm(desc="XGBoost Hyperparameter Search", total=total_fits)):
    start = time.perf_counter()
    search.fit(X_train, y_train,eval_set=[(X_val, y_val)])
    train_time = time.perf_counter() - start

best_xgb = search.best_estimator_
```

```python
y_val_prob = best_xgb.predict_proba(X_val)[:, 1]
val_auc = roc_auc_score(y_val, y_val_prob)

print(f"Validation ROC-AUC: {val_auc:.4f}")
print("Best params:", search.best_params_)
```

```python
start = time.perf_counter()
y_test_prob = best_xgb.predict_proba(X_test)[:, 1]
score_time = time.perf_counter() - start

test_auc = roc_auc_score(y_test, y_test_prob)
latency_ms = (score_time / len(X_test)) * 1_000

print(f"Test ROC-AUC: {test_auc:.4f}")
print(f"Latency per sample: {latency_ms:.3f} ms")
```

```python
y_train_prob = best_xgb.predict_proba(X_train)[:, 1]
train_auc = roc_auc_score(y_train, y_train_prob)

# Test predictions
y_test_prob = best_xgb.predict_proba(X_test)[:, 1]
test_auc = roc_auc_score(y_test, y_test_prob)

print(f"Train ROC-AUC: {train_auc:.4f}")
print(f"Test  ROC-AUC: {test_auc:.4f}")
```

## LGBM

```python
import lightgbm as lgb
from sklearn.model_selection import StratifiedKFold
```

```python
scale_pos_weight = (y == 0).sum() / (y == 1).sum()

lgbm_fs = lgb.LGBMClassifier(
    objective="binary",
    n_estimators=300,
    learning_rate=0.05,
    num_leaves=31,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight,
    random_state=42,
    n_jobs=-1
)
```

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

importance_sum = np.zeros(X.shape[1])

for tr_idx, va_idx in cv.split(X, y):
    lgbm_fs.fit(X.iloc[tr_idx], y.iloc[tr_idx])
    importance_sum += lgbm_fs.booster_.feature_importance(importance_type="gain")

importance = pd.Series(
    importance_sum / cv.get_n_splits(),
    index=X.columns
).sort_values(ascending=False)
```

```python
top_50_features = importance.head(50).index.tolist()
top_50_features
```

```python
X_top50 = all_feats_shrink[[ID_COL, TARGET] + top_50_features]
```

```python
scale_pos_weight = (y_train == 0).sum() / (y_train == 1).sum()

lgbm_model = lgb.LGBMClassifier(
    objective="binary",
    metric="auc",
    scale_pos_weight=scale_pos_weight,
    boosting_type="gbdt",
    random_state=42,
    n_jobs=-1
)

param_dist = {
    "n_estimators": [300, 500, 700],
    "learning_rate": [0.01, 0.02, 0.03],
    "num_leaves": [15, 31, 63],
    "max_depth": [-1, 3, 5, 7],
    "min_child_samples": [20, 50, 100],
    "subsample": [0.6, 0.7, 0.8],
    "colsample_bytree": [0.6, 0.7, 0.8],
    "reg_alpha": [0.0, 0.5, 1.0],
    "reg_lambda": [0.0, 5.0, 10.0]
}
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

search = RandomizedSearchCV(
    estimator=lgbm_model,
    param_distributions=param_dist,
    n_iter=50,
    scoring="roc_auc",
    cv=cv,
    n_jobs=-1,
    random_state=42,
    verbose=0
)

total_fits = 50 * cv.get_n_splits()

with tqdm_joblib(tqdm(desc="LGBM Hyperparameter Search", total=total_fits)):
    start = time.perf_counter()
    search.fit(X_train, y_train)
    train_time = time.perf_counter() - start

best_lgbm = search.best_estimator_
```

```python
y_train_prob = best_lgbm.predict_proba(X_train)[:, 1]
train_auc = roc_auc_score(y_train, y_train_prob)

# Validation
y_val_prob = best_lgbm.predict_proba(X_val)[:, 1]
val_auc = roc_auc_score(y_val, y_val_prob)

# Test
y_test_prob = best_lgbm.predict_proba(X_test)[:, 1]
test_auc = roc_auc_score(y_test, y_test_prob)

print(f"Train ROC-AUC     : {train_auc:.4f}")
print(f"Validation ROC-AUC: {val_auc:.4f}")
print(f"Test ROC-AUC      : {test_auc:.4f}")
```
